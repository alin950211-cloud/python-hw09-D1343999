## 作業說明
這次作業使用 Python 整理兩場活動的報名名單，
透過字串、串列、集合及字典等資料型別，
計算報名人數、人次與活動收入

```python
# 測試一：原始活動報名資料

# 使用 tuple 儲存兩個活動名稱
activity_names = ("桌遊體驗", "電影欣賞")

# 原始報名名單
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 使用 dict 儲存每場活動的報名費
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 整理姓名：移除空白、轉小寫，再用逗號分開
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

print("桌遊報名名單：", board_list)
print("電影報名名單：", movie_list)
```

```python
# HW09：整理活動報名名單

# 使用 set 去除重複報名者
board_set = set(board_list)
movie_set = set(movie_list)

# 找出兩場活動都有參加的人
both = board_set & movie_set

# 找出至少參加一場活動的人
all_people = board_set | movie_set

# 計算不重複人數與總報名人次
unique_count = len(all_people)
total_entries = len(board_set) + len(movie_set)

# 計算兩場活動的報名費
board_income = len(board_set) * unit_fees[activity_names[0]]
movie_income = len(movie_set) * unit_fees[activity_names[1]]
total_income = board_income + movie_income

# 顯示整理結果
print("桌遊體驗名單：", sorted(board_set))
print("電影欣賞名單：", sorted(movie_set))
print("兩場都有參加：", sorted(both))
print("不重複總人數：", unique_count)
print("總報名人次：", total_entries)
print("總應收金額：", total_income, "元")
```

```python
# 找出只參加其中一場活動的人
only_board = board_set - movie_set
only_movie = movie_set - board_set

# 將名單排序
sorted_people = sorted(all_people)

# 使用切片取得前3位通知名單
notify_list = sorted_people[:3]

# 使用字典儲存活動摘要
summary = {
    "不重複人數": unique_count,
    "報名人次": total_entries,
    "總應收金額": total_income
}

print("只參加桌遊：", sorted(only_board))
print("只參加電影：", sorted(only_movie))
print("通知名單：", notify_list)
print("活動摘要：", summary)
```

```python
# 測試二：新增 Frank 參加電影欣賞

# 複製原本名單，避免影響測試一
board_test2 = board_set.copy()
movie_test2 = movie_set.copy()

# 電影欣賞新增 Frank
movie_test2.add("frank")

# 計算不重複人數
all_test2 = board_test2 | movie_test2
unique_test2 = len(all_test2)

# 計算報名人次
entries_test2 = len(board_test2) + len(movie_test2)

# 計算總應收金額
income_test2 = (
    len(board_test2) * unit_fees[activity_names[0]]
    + len(movie_test2) * unit_fees[activity_names[1]]
)

print("測試二：新增 Frank")
print("不重複總人數：", unique_test2)
print("總報名人次：", entries_test2)
print("總應收金額：", income_test2, "元")
```

```python
# 測試三：兩場活動的報名名單完全相同

# 設定兩場活動都是相同的三個人
board_test3 = {"amy", "ben", "cora"}
movie_test3 = {"amy", "ben", "cora"}

# 找出兩場都有參加的人
both_test3 = board_test3 & movie_test3

# 計算不重複人數
all_test3 = board_test3 | movie_test3
unique_test3 = len(all_test3)

# 計算總報名人次
entries_test3 = len(board_test3) + len(movie_test3)

# 計算總應收金額
income_test3 = (
    len(board_test3) * unit_fees[activity_names[0]]
    + len(movie_test3) * unit_fees[activity_names[1]]
)

# 顯示結果
print("測試三：兩場名單完全相同")
print("共同參加者：", sorted(both_test3))
print("不重複總人數：", unique_test3)
print("總報名人次：", entries_test3)
print("總應收金額：", income_test3, "元")
```
